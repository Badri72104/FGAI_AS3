```python
!pip -q install transformers datasets sentencepiece sacrebleu accelerate bitsandbytes pandas scikit-learn

import os, time, json, random
import numpy as np
import pandas as pd
import torch
import sacrebleu

from datasets import load_dataset
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", DEVICE)
if DEVICE == "cuda":
    print(torch.cuda.get_device_name(0))

# New project directory
from google.colab import drive
drive.mount("/content/drive")

PROJECT = "/content/drive/MyDrive/FGAI_Assignment3_Simple"
os.makedirs(PROJECT + "/models", exist_ok=True)
os.makedirs(PROJECT + "/results", exist_ok=True)

# Load English-French OPUS Books
raw = load_dataset("Helsinki-NLP/opus_books", "en-fr", split="train")

pairs = []
for row in raw:
    translation = row["translation"]
    en = translation.get("en", "").strip()
    fr = translation.get("fr", "").strip()

    if en and fr and len(en) > 5 and len(fr) > 5:
        pairs.append({"source": en, "target": fr})

df = pd.DataFrame(pairs).drop_duplicates()
df = df.sample(n=min(2500, len(df)), random_state=SEED)

# Explicit 80/10/10 split
train_df, temp_df = train_test_split(
    df, test_size=0.20, random_state=SEED
)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50, random_state=SEED
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

train_df.to_csv(PROJECT + "/train.csv", index=False)
val_df.to_csv(PROJECT + "/validation.csv", index=False)
test_df.to_csv(PROJECT + "/test.csv", index=False)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
print("\nExample:")
print(train_df.iloc[0].to_dict())

# Shared tokenizer for the scratch Transformer
tokenizer = AutoTokenizer.from_pretrained("google/mt5-small")
VOCAB_SIZE = len(tokenizer)
PAD_ID = tokenizer.pad_token_id
EOS_ID = tokenizer.eos_token_id

print("Tokenizer vocabulary:", VOCAB_SIZE)
```

```python
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

MAX_LEN = 64
D_MODEL = 128

class TranslationDataset(Dataset):
    def __init__(self, frame):
        self.src = []
        self.tgt = []

        for _, row in frame.iterrows():
            s = tokenizer(
                row.source, truncation=True, max_length=MAX_LEN-1
            )["input_ids"] + [EOS_ID]

            t = tokenizer(
                row.target, truncation=True, max_length=MAX_LEN-1
            )["input_ids"] + [EOS_ID]

            self.src.append(s)
            self.tgt.append(t)

    def __len__(self):
        return len(self.src)

    def __getitem__(self, i):
        return self.src[i], self.tgt[i]


def collate(batch):
    src, tgt = zip(*batch)

    src = nn.utils.rnn.pad_sequence(
        [torch.tensor(x) for x in src],
        batch_first=True, padding_value=PAD_ID
    )
    tgt = nn.utils.rnn.pad_sequence(
        [torch.tensor(x) for x in tgt],
        batch_first=True, padding_value=PAD_ID
    )
    return src.to(DEVICE), tgt.to(DEVICE)


class BasicTransformer(nn.Module):
    def __init__(self):
        super().__init__()
        self.embedding = nn.Embedding(VOCAB_SIZE, D_MODEL, padding_idx=PAD_ID)
        self.position = nn.Embedding(MAX_LEN, D_MODEL)

        self.transformer = nn.Transformer(
            d_model=D_MODEL,
            nhead=4,
            num_encoder_layers=2,
            num_decoder_layers=2,
            dim_feedforward=256,
            dropout=0.1,
            batch_first=True
        )

        self.output = nn.Linear(D_MODEL, VOCAB_SIZE, bias=False)
        self.output.weight = self.embedding.weight

    def forward(self, src, tgt):
        src_pos = torch.arange(src.size(1), device=DEVICE)
        tgt_pos = torch.arange(tgt.size(1), device=DEVICE)

        src_emb = self.embedding(src) + self.position(src_pos)
        tgt_emb = self.embedding(tgt) + self.position(tgt_pos)

        causal_mask = torch.triu(
            torch.ones(tgt.size(1), tgt.size(1), device=DEVICE),
            diagonal=1
        ).bool()

        output = self.transformer(
            src_emb, tgt_emb,
            tgt_mask=causal_mask,
            src_key_padding_mask=(src == PAD_ID),
            tgt_key_padding_mask=(tgt == PAD_ID),
            memory_key_padding_mask=(src == PAD_ID)
        )
        return self.output(output)


basic = BasicTransformer().to(DEVICE)

train_loader = DataLoader(
    TranslationDataset(train_df),
    batch_size=32, shuffle=True, collate_fn=collate
)

val_loader = DataLoader(
    TranslationDataset(val_df),
    batch_size=32, shuffle=False, collate_fn=collate
)

optimizer = torch.optim.Adam(basic.parameters(), lr=0.0003)

start = time.time()
best_val = float("inf")

for epoch in range(3):
    basic.train()
    train_losses = []

    for src, tgt in train_loader:
        optimizer.zero_grad()

        logits = basic(src, tgt[:, :-1])
        loss = F.cross_entropy(
            logits.reshape(-1, VOCAB_SIZE),
            tgt[:, 1:].reshape(-1),
            ignore_index=PAD_ID
        )

        loss.backward()
        torch.nn.utils.clip_grad_norm_(basic.parameters(), 1.0)
        optimizer.step()
        train_losses.append(loss.item())

    basic.eval()
    val_losses = []

    with torch.no_grad():
        for src, tgt in val_loader:
            logits = basic(src, tgt[:, :-1])
            loss = F.cross_entropy(
                logits.reshape(-1, VOCAB_SIZE),
                tgt[:, 1:].reshape(-1),
                ignore_index=PAD_ID
            )
            val_losses.append(loss.item())

    val_loss = np.mean(val_losses)
    print(
        f"Epoch {epoch+1}/3 | "
        f"Train: {np.mean(train_losses):.4f} | "
        f"Validation: {val_loss:.4f}"
    )

    if val_loss < best_val:
        best_val = val_loss
        torch.save(basic.state_dict(), PROJECT + "/models/basic.pt")

basic_time = time.time() - start
basic.load_state_dict(torch.load(
    PROJECT + "/models/basic.pt", map_location=DEVICE
))
basic.eval()

# Greedy translation
def basic_translate(text):
    src = tokenizer(
        text, return_tensors="pt",
        truncation=True, max_length=MAX_LEN
    )["input_ids"].to(DEVICE)

    generated = torch.tensor([[PAD_ID]], device=DEVICE)

    with torch.no_grad():
        for _ in range(MAX_LEN):
            logits = basic(src, generated)
            next_token = logits[:, -1, :].argmax(-1, keepdim=True)
            generated = torch.cat([generated, next_token], dim=1)

            if next_token.item() == EOS_ID:
                break

    return tokenizer.decode(
        generated[0, 1:], skip_special_tokens=True
    )

basic_predictions = [basic_translate(x) for x in test_df.source]
```

```python
import torch
import torch.nn as nn
import pandas as pd
import time
import sacrebleu
from transformers import AutoTokenizer

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("google/mt5-small")

PAD_ID = tokenizer.pad_token_id
EOS_ID = tokenizer.eos_token_id
VOCAB_SIZE = len(tokenizer)

D_MODEL = 128
MAX_LEN = 64

# Reconstruct exact architecture
class BasicTransformer(nn.Module):
    def __init__(self):
        super().__init__()

        self.embedding = nn.Embedding(VOCAB_SIZE, D_MODEL)
        self.position = nn.Embedding(MAX_LEN, D_MODEL)

        self.transformer = nn.Transformer(
            d_model=D_MODEL,
            nhead=4,
            num_encoder_layers=2,
            num_decoder_layers=2,
            dim_feedforward=256,
            batch_first=True
        )

        self.output = nn.Linear(D_MODEL, VOCAB_SIZE, bias=False)

    def forward(self, src, tgt):
        src_pos = torch.arange(src.size(1), device=src.device)
        tgt_pos = torch.arange(tgt.size(1), device=tgt.device)

        src_emb = self.embedding(src) + self.position(src_pos)
        tgt_emb = self.embedding(tgt) + self.position(tgt_pos)

        causal_mask = nn.Transformer.generate_square_subsequent_mask(
            tgt.size(1), device=tgt.device
        )

        out = self.transformer(
            src_emb,
            tgt_emb,
            tgt_mask=causal_mask,
            src_key_padding_mask=(src == PAD_ID),
            tgt_key_padding_mask=(tgt == PAD_ID),
            memory_key_padding_mask=(src == PAD_ID)
        )

        return self.output(out)

# Load checkpoint
basic = BasicTransformer().to(DEVICE)

state = torch.load(
    f"{PROJECT}/models/basic.pt",
    map_location=DEVICE,
    weights_only=True
)

basic.load_state_dict(state)
basic.eval()

print("Basic Transformer loaded successfully.")

# Greedy decoding
def translate(sentence):
    src = tokenizer(
        sentence,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LEN
    )["input_ids"].to(DEVICE)

    generated = torch.tensor([[PAD_ID]], device=DEVICE)

    with torch.no_grad():
        for _ in range(MAX_LEN - 1):
            logits = basic(src, generated)
            next_token = logits[:, -1, :].argmax(dim=-1).item()

            if next_token == EOS_ID:
                break

            generated = torch.cat([
                generated,
                torch.tensor([[next_token]], device=DEVICE)
            ], dim=1)

    return tokenizer.decode(
        generated[0, 1:],
        skip_special_tokens=True
    )

# Evaluate
test_df = pd.read_csv(f"{PROJECT}/test.csv").head(250)

references = test_df["target"].astype(str).tolist()
predictions = []

torch.cuda.synchronize()
start = time.time()

for i, source in enumerate(test_df["source"].astype(str)):
    predictions.append(translate(source))

    if (i + 1) % 25 == 0:
        print(f"Processed {i+1}/250")

torch.cuda.synchronize()
inference_time = time.time() - start

print("\nBASIC TRANSFORMER RESULTS")
print("BLEU:", sacrebleu.corpus_bleu(
    predictions, [references]
).score)

print("chrF:", sacrebleu.corpus_chrf(
    predictions, [references]
).score)

print("Total inference time:", inference_time)
print("Average inference:", inference_time / len(predictions))

pd.DataFrame({
    "source": test_df["source"],
    "reference": references,
    "prediction": predictions
}).to_csv(
    f"{PROJECT}/results/basic_predictions.csv",
    index=False
)

print("Predictions saved.")
```
